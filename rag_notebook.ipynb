```python
from langchain_community.document_loaders import PyPDFLoader, TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter  # ✅ 修改这里
import os

print("✅ 包导入成功")
```

```python
def load_documents(folder_path: str):
    docs = []
    for filename in os.listdir(folder_path):
        file_full_path = os.path.join(folder_path, filename)
        if filename.endswith(".pdf"):
            loader = PyPDFLoader(file_full_path)
            docs.extend(loader.load())
        elif filename.endswith(".txt"):
            loader = TextLoader(file_full_path, encoding="utf-8")
            docs.extend(loader.load())
    return docs

def split_documents(documents, chunk_size=800, chunk_overlap=120):
    splitter = RecursiveCharacterTextSplitter(
        chunk_size=chunk_size,
        chunk_overlap=chunk_overlap,
        length_function=len
    )
    chunks = splitter.split_documents(documents)
    print(f"原始文档切分为 {len(chunks)} 个文本块")
    return chunks
```

```python
raw_docs = load_documents("./knowledge")
chunk_result = split_documents(raw_docs)
print("第一个文本块内容：")
print(chunk_result[0].page_content)
```
